```python
# Define the folder path
folder_path = 'data/tests'

# Loop through each file in the folder
for file_path in glob.glob(os.path.join(folder_path, '*')):
    with open(file_path) as file:
        # Process the file
        csv_as_dataframe = pd.read_csv(file)
        # csv_as_dataframe.info()
        report = classification_report(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"])
        print(f"Results Task A {file_path.split('/')[-1]}")
        
        metrics_df = pd.DataFrame(columns=['accuracy', 'precision', 'recall', 'f1', 'roc_auc'])

        print("Confusion Matrix:")
        print(confusion_matrix(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"]))
        
        accuracy = (csv_as_dataframe["target_true"] == csv_as_dataframe["target_pred"]).mean()
        precision = precision_score(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"])
        recall = recall_score(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"])
        f1 = f1_score(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"])
        roc_auc = roc_auc_score(csv_as_dataframe["target_true"].astype(int), csv_as_dataframe["target_pred"])
        metrics_df.loc[0] = [accuracy, precision, recall, f1, roc_auc]
        print("\nMetrics:")
        print(metrics_df)
        print()
        print()
```
